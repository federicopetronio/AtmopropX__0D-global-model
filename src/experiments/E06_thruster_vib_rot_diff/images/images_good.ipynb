```python
import scipy.constants as cons
from datetime import datetime
from nrlmsise00 import msise_model
 
def entering_flux(altitude, AREA = 1.0):
    '''Calculate injection rates for O2, N2, O, N at a given altitude (in kilometers) and area (in m^2) using the MSISE-00 model.'''
    # Mass of species in kg
    M_O_Kg = 16 * cons.atomic_mass  # kg
    M_O2_Kg = 32 * cons.atomic_mass  # kg
    M_N_Kg = 14 * cons.atomic_mass  # kg
    M_N2_Kg = 28 * cons.atomic_mass  # kg
    print("Calculating injection rates at altitude:", altitude, "km")

    # dens = msise_flat(datetime(2025, 1, 1, 0, 0, 0), np.array([[altitude]]), 0, -70, 150, 150, 4)  # convert from cm^-3 to m^-3
    # O2_den = dens[0,0,3] * 1e6  # convert from cm^-3 to m^-3
    # N2_den = dens[0,0,2] * 1e6  # convert from cm^-3 to m^-3
    # O_den = dens[0,0,1] * 1e6  # convert from cm^-3 to m^-3
    # N_den = dens[0,0,7] * 1e6  # convert from cm^-3 to m^-3
    # T_eff = 0.07
    
    dens, temp = msise_model(datetime(2025, 1, 1, 0, 0, 0), np.array([[altitude]]), 0, 0, 150, 150, 4)

    O2_den = dens[3] * 1e6  # convert from cm^-3 to m^-3
    N2_den = dens[2] * 1e6  # convert from cm^-3 to m^-3
    O_den = dens[1] * 1e6  # convert from cm^-3 to m^-3
    N_den = dens[7] * 1e6  # convert from cm^-3 to m^-3
    T_eff = temp[1] / 11604.52500617  # convert from K to eV
    T_eff = 0.03

    mu = 3.986e14  # m^3/s^2
    R_earth = 6371e3  # m
    orbital_speeds = np.sqrt(mu / (R_earth + altitude*1e3))

    O2_injection_rate = orbital_speeds * np.array(O2_den) * AREA  # particles/s
    N2_injection_rate = orbital_speeds * np.array(N2_den) * AREA  # particles/s
    O_injection_rate = orbital_speeds * np.array(O_den) * AREA  # particles/s
    N_injection_rate = orbital_speeds * np.array(N_den) * AREA  # particles/s

    return O2_injection_rate * M_O2_Kg, N2_injection_rate * M_N2_Kg, O_injection_rate * M_O_Kg, N_injection_rate * M_N_Kg
```

```python
# child-langmuir limit calculation
import scipy.constants as cons
M_O_Kg = 16 * cons.atomic_mass  # kg
M_O2_Kg = 32 * cons.atomic_mass  # kg
M_N_Kg = 14 * cons.atomic_mass  # kg
M_N2_Kg = 28 * cons.atomic_mass  # kg
M_Xe_Kg = 131.293 * cons.atomic_mass  # kg
M_I_Kg = 126.90447 * cons.atomic_mass  # kg

V_grid = 1000  # Volts
ss = 1e-3 # m

J_CL_O = 4/9*cons.epsilon_0*np.sqrt(2*cons.e/(M_O_Kg))* (V_grid)**1.5 / (ss)**2
print(f"Child-Langmuir current density for O+ at V={V_grid} V and s={ss*1e3} mm gap: J_CL_O = {J_CL_O:.2f} A/m2")
J_CL_O2 = 4/9*cons.epsilon_0*np.sqrt(2*cons.e/(M_O2_Kg))* (V_grid)**1.5 / (ss)**2
print(f"Child-Langmuir current density for O2+ at V={V_grid} V and s={ss*1e3} mm gap: J_CL_O2 = {J_CL_O2:.2f} A/m2")
J_CL_N = 4/9*cons.epsilon_0*np.sqrt(2*cons.e/(M_N_Kg))* (V_grid)**1.5 / (ss)**2
print(f"Child-Langmuir current density for N+ at V={V_grid} V and s={ss*1e3} mm gap: J_CL_N = {J_CL_N:.2f} A/m2")
J_CL_N2 = 4/9*cons.epsilon_0*np.sqrt(2*cons.e/(M_N2_Kg))* (V_grid)**1.5 / (ss)**2
print(f"Child-Langmuir current density for O+ at V={V_grid} V and s={ss*1e3} mm gap: J_CL_O = {J_CL_O:.2f} A/m2")
J_CL_Xe = 4/9*cons.epsilon_0*np.sqrt(2*cons.e/(M_Xe_Kg))* (V_grid)**1.5 / (ss)**2
print(f"Child-Langmuir current density for Xe+ at V={V_grid} V and s={ss*1e3} mm gap: J_CL_Xe = {J_CL_Xe:.2f} A/m2")
J_CL_I = 4/9*cons.epsilon_0*np.sqrt(2*cons.e/(M_I_Kg))* (V_grid)**1.5 / (ss)**2
print(f"Child-Langmuir current density for I+ at V={V_grid} V and s={ss*1e3} mm gap: J_CL_I = {J_CL_I:.2f} A/m2")
```

```python
SIGMA_I = 1e-18  # m2
L = 10e-2  # m

def u_B(T_e, m_ion):
    """T_e in eV, m_ion is mass of single ion"""
    return np.sqrt(cons.e*T_e/(m_ion))

def h_L(n_g_tot):
    lambda_i = 1/(n_g_tot * SIGMA_I)
    return 0.86 / np.sqrt(3 + (L / (2 * lambda_i)))
```

```python
## ion current

n_g_tot = density_O + density_O2 + density_N + density_N2
h_L_calc = h_L(n_g_tot)

# O
ni = density_Op
u_B_val = u_B(Temp_e, M_O_Kg)
J_ion_O = ni * cons.e * u_B_val * h_L_calc

#O2
ni_O2 = density_O2p
u_B_val_O2 = u_B(Temp_e, M_O2_Kg)
J_ion_O2 = ni_O2 * cons.e * u_B_val_O2 * h_L_calc

# N
ni_N = density_Np
u_B_val_N = u_B(Temp_e, M_N_Kg)
J_ion_N = ni_N * cons.e * u_B_val_N * h_L_calc

# N2
ni_N2 = density_N2p
u_B_val_N2 = u_B(Temp_e, M_N2_Kg)
J_ion_N2 = ni_N2 * cons.e * u_B_val_N2 * h_L_calc

print(f"Ion current density for O+ at P={Power.iloc[-1]} W: J_ion_O = {J_ion_O.iloc[-1]:.2f} A/m2")
fig = plt.figure(figsize=(image_width_large, 6))
gs = gridspec.GridSpec(2, 2)
gs.update(wspace=0.4, hspace=0.4)
ax0 = fig.add_subplot(gs[0, 0])
ax1 = fig.add_subplot(gs[0, 1])
ax2 = fig.add_subplot(gs[1, 0])
ax3 = fig.add_subplot(gs[1, 1])

ax0.plot(Power, J_ion_O, label='Ion current density O+', marker='x')
ax0.axhline(J_CL_O, color='r', linestyle='--', label='Child-Langmuir limit O+')
ax0.set_xlabel('RF Power [W]')
ax0.set_ylabel('Ion current density [A/m2]')
ax0.set_yscale('log')
ax0.set_title('Ion current density vs RF Power for O+', fontsize=10)

ax1.plot(Power, J_ion_O2, label='Ion current density O2+', marker='x')
ax1.axhline(J_CL_O2, color='r', linestyle='--', label='Child-Langmuir limit O+')
ax1.set_xlabel('RF Power [W]')
ax1.set_ylabel('Ion current density [A/m2]')
ax1.set_yscale('log')
ax1.set_title('Ion current density vs RF Power for O2+', fontsize=10)

ax2.plot(Power, J_ion_N, label='Ion current density N+', marker='x')
ax2.axhline(J_CL_N, color='r', linestyle='--', label='Child-Langmuir limit N+')
ax2.set_xlabel('RF Power [W]')
ax2.set_ylabel('Ion current density [A/m2]')
ax2.set_yscale('log')
ax2.set_title('Ion current density vs RF Power for N+', fontsize=10) 

ax3.plot(Power, J_ion_N2, label='Ion current density N2+', marker='x')
ax3.axhline(J_CL_N2, color='r', linestyle='--', label='Child-Langmuir limit N2+')
ax3.set_xlabel('RF Power [W]')
ax3.set_ylabel('Ion current density [A/m2]')
ax3.set_yscale('log')
ax3.set_title('Ion current density vs RF Power for N2+', fontsize=10)

```

```python
from nrlmsise00 import msise_model
from datetime import datetime

def calc_inection_rate_air(altitude, AREA = 1.0):
    '''Calculate injection rates for O2, N2, O, N at a given altitude (in kilometers) and area (in m^2) using the MSISE-00 model.'''
    # print("Calculating injection rates at altitude:", altitude, "km")

    # dens = msise_flat(datetime(2025, 1, 1, 0, 0, 0), np.array([[altitude]]), 0, -70, 150, 150, 4)  # convert from cm^-3 to m^-3
    # O2_den = dens[0,0,3] * 1e6  # convert from cm^-3 to m^-3
    # N2_den = dens[0,0,2] * 1e6  # convert from cm^-3 to m^-3
    # O_den = dens[0,0,1] * 1e6  # convert from cm^-3 to m^-3
    # N_den = dens[0,0,7] * 1e6  # convert from cm^-3 to m^-3
    # T_eff = 0.07
    
    dens, temp = msise_model(datetime(2025, 1, 1, 0, 0, 0), np.array([[altitude]]), 0, 0, 150, 150, 4)

    O2_den = dens[3] * 1e6  # convert from cm^-3 to m^-3
    N2_den = dens[2] * 1e6  # convert from cm^-3 to m^-3
    O_den = dens[1] * 1e6  # convert from cm^-3 to m^-3
    N_den = dens[7] * 1e6  # convert from cm^-3 to m^-3
    T_eff = temp[1] / 11604.52500617  # convert from K to eV
    T_eff = 0.03

    mu = 3.986e14  # m^3/s^2
    R_earth = 6371e3  # m
    orbital_speeds = np.sqrt(mu / (R_earth + altitude*1e3))

    O2_injection_rate = orbital_speeds * np.array(O2_den) * AREA  # particles/s
    N2_injection_rate = orbital_speeds * np.array(N2_den) * AREA  # particles/s
    O_injection_rate = orbital_speeds * np.array(O_den) * AREA  # particles/s
    N_injection_rate = orbital_speeds * np.array(N_den) * AREA  # particles/s
    dframe = pd.DataFrame({
        "Heit(km)": [altitude],
        "Oden(m-3)": [O_den],
        "O2den(m-3)": [O2_den],
        "Nden(m-3)": [N_den],
        "N2den(m-3)": [N2_den],
        "T(eV)": [T_eff],
        "Q_O2(s-1)": [O2_injection_rate],
        "Q_N2(s-1)": [N2_injection_rate],
        "Q_O(s-1)": [O_injection_rate],
        "Q_N(s-1)": [N_injection_rate],
    })
    return dframe

#1500 W
# df_results.to_csv(f"data/results_scan_altitude_{power}.csv", index=False)
df_1500 = pd.read_csv("../data/results_scan_altitude_1500.csv") # etaC=0.5
# df_1500_1 = pd.read_csv("../data/results_scan_altitude_1500_etaC=0.35_1.csv") # etaC=0.35
df_1500_1 = pd.read_csv("../data/results_scan_altitude_1500_etaC=0.35.csv") # etaC=0.35
df_2000 = pd.read_csv("../data/results_scan_altitude_2000.csv") # etaC=0.5
# df_2000_1 = pd.read_csv("../data/results_scan_altitude_2000_etaC=0.35_1.csv") # etaC=0.35
df_2000_1 = pd.read_csv("../data/results_scan_altitude_2000_etaC=0.35.csv") # etaC=0.35
df_2500 = pd.read_csv("../data/results_scan_altitude_2500.csv") # etaC=0.5
# df_2500_1 = pd.read_csv("../data/results_scan_altitude_2500_etaC=0.35_1.csv") # etaC=0.35
df_2500_1 = pd.read_csv("../data/results_scan_altitude_2500_etaC=0.35.csv") # etaC=0.35
df_3000 = pd.read_csv("../data/results_scan_altitude_3000.csv") # etaC=0.5
# df_3000_1 = pd.read_csv("../data/results_scan_altitude_3000_etaC=0.35_1.csv") # etaC=0.35
df_3000_1 = pd.read_csv("../data/results_scan_altitude_3000_etaC=0.35.csv") # etaC=0.35


# # add a line at altitude 300 km with zeros to df_3000_1
# for df in [df_2500_1, df_3000_1]:
#     df = pd.concat([df, pd.DataFrame({
#         "Power (W)": [3000],
#         "n_e (m-3)": [0],
#         "n_O (m-3)": [0],
#         "n_O2 (m-3)": [0],
#         "n_O+ (m-3)": [0],
#         "n_O2+ (m-3)": [0],
#         "n_N (m-3)": [0],
#         "n_N2 (m-3)": [0],
#         "n_N+ (m-3)": [0],
#         "n_N2+ (m-3)": [0],
#         "T_e (eV)": [0],
#         "T_ng_mono (eV)": [0],
#         "T_ng_diato (eV)": [0],
#         "Thrust_ion (N)": [0],
#         "Thrust_neutral (N)": [0],
#         "Power_transfer_efficiency": [0],
#         "Gamma_efficiency": [0],
#         "eta_efficiency": [0],
#         "xi_efficiency": [0]
#     })], ignore_index=True)

```

```python
from nrlmsise00 import msise_model
from datetime import datetime

def calc_inection_rate_air(altitude, AREA = 1.0):
    '''Calculate injection rates for O2, N2, O, N at a given altitude (in kilometers) and area (in m^2) using the MSISE-00 model.'''
    print("Calculating injection rates at altitude:", altitude, "km")

    # dens = msise_flat(datetime(2025, 1, 1, 0, 0, 0), np.array([[altitude]]), 0, -70, 150, 150, 4)  # convert from cm^-3 to m^-3
    # O2_den = dens[0,0,3] * 1e6  # convert from cm^-3 to m^-3
    # N2_den = dens[0,0,2] * 1e6  # convert from cm^-3 to m^-3
    # O_den = dens[0,0,1] * 1e6  # convert from cm^-3 to m^-3
    # N_den = dens[0,0,7] * 1e6  # convert from cm^-3 to m^-3
    # T_eff = 0.07
    
    dens, temp = msise_model(datetime(2025, 1, 1, 0, 0, 0), np.array([[altitude]]), 0, 0, 150, 150, 4)
    
    O2_den = dens[3] * 1e6  # convert from cm^-3 to m^-3
    N2_den = dens[2] * 1e6  # convert from cm^-3 to m^-3
    O_den = dens[1] * 1e6  # convert from cm^-3 to m^-3
    N_den = dens[7] * 1e6  # convert from cm^-3 to m^-3
    He_den = dens[0] * 1e6  # convert from cm^-3 to m^-3
    Ar_den = dens[4] * 1e6  # convert from cm^-3 to m^-3
    T_eff = temp[1] / 11604.52500617  # convert from K to eV

    mu = 3.986e14  # m^3/s^2
    R_earth = 6371e3  # m
    orbital_speeds = np.sqrt(mu / (R_earth + altitude*1e3))

    O2_injection_rate = orbital_speeds * np.array(O2_den) * AREA  # particles/s
    N2_injection_rate = orbital_speeds * np.array(N2_den) * AREA  # particles/s
    O_injection_rate = orbital_speeds * np.array(O_den) * AREA  # particles/s
    N_injection_rate = orbital_speeds * np.array(N_den) * AREA  # particles/s
    dframe = pd.DataFrame({
        "Heit(km)": [altitude],
        "Oden(m-3)": [O_den],
        "O2den(m-3)": [O2_den],
        "Nden(m-3)": [N_den],
        "N2den(m-3)": [N2_den],
        "T(eV)": [T_eff],
        "Q_O2(s-1)": [O2_injection_rate],
        "Q_N2(s-1)": [N2_injection_rate],
        "Q_O(s-1)": [O_injection_rate],
        "Q_N(s-1)": [N_injection_rate],
        "Heden(m-3)": [He_den],
        "Arden(m-3)": [Ar_den],
    })
    return dframe
```
